```python
import pandas as pd
pd.options.display.float_format = '{:,.2f}'.format
import numpy as np
```

```python
dados = "C:/Users/xavier.thiago/Documents/Projeto_Analise_Dados_Caramelo_Store_SENAC_RJ/data/CARAMELO.csv"
df = pd.read_csv(dados, sep=';', encoding='UTF-8')
df.info()
```

```python
df['Valor de Venda'] = df['Valor de Venda'].str.replace(',', '.').astype(float)
df.info()
```

```python
array_venda = np.array(df['Valor de Venda'])
q1 = np.percentile(array_venda, 25)
q2 = np.percentile(array_venda, 50)
q3 = np.percentile(array_venda, 75)
print(f'Primeiro quartil (Q1): {q1}')
print(f'Segundo quartil (Q2, Mediana): {q2}')
print(f'Terceiro quartil (Q3): {q3}')
```

Valor de venda por loja

```python
df_valor_venda_loja = df
df_valor_venda_loja = df_valor_venda_loja.groupby(['Loja'])['Valor de Venda'].sum().reset_index().sort_values(by='Valor de Venda', ascending=False)
df_valor_venda_loja
```

```python
array_valor_venda_loja = np.array(df_valor_venda_loja['Valor de Venda'])

q1_valor_venda_loja = np.percentile(array_valor_venda_loja, 25)
q2_valor_venda_loja = np.percentile(array_valor_venda_loja, 50)
q3_valor_venda_loja = np.percentile(array_valor_venda_loja, 75)

media_valor_venda_loja = np.mean(array_valor_venda_loja)
mediana_valor_venda_loja = np.median(array_valor_venda_loja)
distancia_valor_venda_loja = (media_valor_venda_loja - mediana_valor_venda_loja) / mediana_valor_venda_loja

iqr_valor_venda_loja = q3_valor_venda_loja - q1_valor_venda_loja
limite_superior_valor_venda_loja = q3_valor_venda_loja + (1.5 * iqr_valor_venda_loja)
limite_inferior_valor_venda_loja = q1_valor_venda_loja - (1.5 * iqr_valor_venda_loja)

print(f'Primeiro quartil (Q1): {q1_valor_venda_loja}')
print(f'Segundo quartil (Q2, Mediana): {q2_valor_venda_loja}')
print(f'Terceiro quartil (Q3): {q3_valor_venda_loja}')

print(f'O valor da média é {media_valor_venda_loja}')
print(f'O valor da mediana é {mediana_valor_venda_loja}')
print(f'A distância entre média e mediana é de {distancia_valor_venda_loja * 100}')

print(f'O limite superior para detecção de outliers é {limite_superior_valor_venda_loja}')
print(f'O limite inferior para detecção de outliers é {limite_inferior_valor_venda_loja}')
```

```python
outliers_valor_venda_loja_limite_inferior = df_valor_venda_loja.loc[df_valor_venda_loja['Valor de Venda'] <= limite_inferior_valor_venda_loja]
outliers_valor_venda_loja_limite_inferior = outliers_valor_venda_loja_limite_inferior.sort_values(by='Valor de Venda')
outliers_valor_venda_loja_limite_inferior
```

```python
outliers_valor_venda_loja_limite_superior = df_valor_venda_loja.loc[df_valor_venda_loja['Valor de Venda'] >= limite_superior_valor_venda_loja]
outliers_valor_venda_loja_limite_superior = outliers_valor_venda_loja_limite_superior.sort_values(by='Valor de Venda')
outliers_valor_venda_loja_limite_superior
```
